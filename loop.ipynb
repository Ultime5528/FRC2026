```python
from wpimath.kinematics import SwerveModuleState
from wpimath.geometry import Rotation2d
import math
```

```python
state = SwerveModuleState(3.2, Rotation2d.fromDegrees(92))
accel_meters_per_sec = 0.0
```

```python
%%timeit

accel_meters_per_sec = 0.0

corrected_desired_state = SwerveModuleState()
corrected_desired_state.speed = state.speed
corrected_desired_state.angle = state.angle.rotateBy(
    Rotation2d(math.pi / 2)
)

current_rotation = Rotation2d(0.1)

if abs((corrected_desired_state.angle - current_rotation).degrees()) > 90.0:
    corrected_desired_state.speed *= -1
    corrected_desired_state.angle += Rotation2d.fromDegrees(180.0)
    accel_meters_per_sec *= -1

corrected_desired_state.speed *= (
    current_rotation - corrected_desired_state.angle
).cos()

a = corrected_desired_state.speed
b = corrected_desired_state.angle
```

```python
pi_over_two = math.pi / 2
pi = math.pi
two_pi = 2 * math.pi

def normalize_angle_degrees(degrees):
    return (degrees + 180) % 360 - 180

def normalize_angle_rad(rad):
    return (rad + pi) % two_pi - pi
```

```python
%%timeit

accel_meters_per_sec = 0.0

speed = state.speed
angle = state.angle.radians() + pi_over_two

current_rotation = 0.1

if abs(normalize_angle_rad(angle - current_rotation)) > pi_over_two:
    speed *= -1
    angle = normalize_angle_rad(angle + pi)
    accel_meters_per_sec *= -1

corrected_desired_state.speed *= math.cos(current_rotation - angle)

a = speed
b = Rotation2d(angle)
```

```python
class Rotation2d:
    @classmethod
    def fromDegrees(cls, degrees):
        return cls(math.radians(degrees))

    @classmethod
    def fromRadians(cls, radians):
        return cls(radians)

    def __init__(self, rad):
        self.radians = rad

    def degrees(self):
        return math.degrees(self.radians)

    def __add__(self, other: Rotation2d):
        return Rotation2d(normalize_angle_rad(self.radians + other.radians))

    def __sub__(self, other: Rotation2d):
        return Rotation2d(normalize_angle_rad(self.radians - other.radians))

    def cos(self):
        return math.cos(self.radians)
```

```python
state_angle = Rotation2d.fromDegrees(92)
offset = Rotation2d(pi_over_two)

from wpimath.geometry import Rotation2d as wpiRotation2d
```

```python
%%timeit

accel_meters_per_sec = 0.0

speed = state.speed
angle = state_angle + offset

current_rotation = Rotation2d(0.1)

if abs((angle - current_rotation).degrees()) > 90.0:
    speed *= -1
    angle += Rotation2d.fromDegrees(180.0)
    accel_meters_per_sec *= -1

speed *= (
    current_rotation - angle
).cos()

a = speed
b = wpiRotation2d(angle.radians)
```
